```python
import torch

logits = torch.rand([2, 3, 4])
labels = torch.rand([2, 3])
_, _, vocab_size = logits.size()
shift_logits = logits[..., :-1, :].contiguous()
shift_labels = labels[..., 1:].contiguous()
shift_masks = (shift_labels < 0.1).to(shift_logits.device)
entropy = (shift_logits.softmax(dim=-1) * shift_logits.log_softmax(dim=-1)).sum(dim=-1)
entropy.masked_fill_(shift_masks, 0.0)
print(entropy)
print(shift_masks)
entropy = entropy.sum(dim=-1) / ((shift_labels >= 0.1).to(shift_logits.device).long().sum(dim=-1) + 1e-8)
extra_shift_masked = torch.logical_or(entropy < 0.1, entropy > 0.2)
shift_labels.masked_fill_(extra_shift_masked.unsqueeze_(-1).expand_as(shift_labels), 0)
print(logits)
print(labels)
print(entropy)
```
